# Question 1:
Use collections.Counter to count word frequencies,
defaultdict(list) to group words by their first letter,
and namedtuple to create lightweight structured records.

```python
from collections import Counter , defaultdict , namedtuple

words = [
    "apple",
    "banana",
    "apple",
    "orange",
    "banana",
    "apple",
    "grape",
    "orange"
]

# 1 counter 
frequency=Counter(words)
print("word frequencies :", frequency)

# 2 default dict
groups =defaultdict(list)
for word in words:
    groups[word[0]].append(word)

print("words groupped by first letter: " , dict(groups))

# 3 namedtuple
word = namedtuple("word" , ["name","length"])
word_info = word("apple",len("apple"))

print("name :",word_info.name)
print("length :" , word_info.length)
```

# Question 2:
Simulate fetching a few items using time.sleep().
First run them one after another, then use ThreadPoolExecutor , and compare the total execution time.

```python
import time
from concurrent.futures import ThreadPoolExecutor

def fetch_item(item):
    print(f"fetching {item}")
    time.sleep(2)
    print(f"finished {item}")
    return item

items = ["Item 1", "Item 2", "Item 3", "Item 4", "Item 5"]

# sequential execution
start = time.perf_counter()
for item in items:
    fetch_item(item)

sequential_time= time.perf_counter()-start

print ("sequential time :" ,sequential_time )

# concurret execution
start = time.perf_counter()

with ThreadPoolExecutor(max_workers=5) as executor:
    results = list(executor.map(fetch_item, items))

concurrent_time = time.perf_counter() - start

print("concurrent time :" , concurrent_time)

#compare results
print ( f"time difference is : {sequential_time-concurrent_time}")
```

# Question 3:
Compute the SHA-256 hash of a string, print it,
then recompute the hash and confirm that both hashes match.

```python
import hashlib
data ="Hello World"

# first hash
hash1 = hashlib.sha256(data.encode()).hexdigest()
print("first sha 256 hash:")
print(hash1)

# recompute the hash
hash2=hashlib.sha256(data.encode()).hexdigest()
print("second sha 256 hash:")
print(hash2)

# compare the hash
if hash1==hash2:
    print("hash is same and the data is unchanged")
else :
    print("hash does not match - data may have changed")
```

# Question 4:
Define a small Pydantic model for a record with name, age, and email.
Create one valid record and one invalid record to observe validation errors.

```python
from pydantic import BaseModel , ValidationError

class User(BaseModel):
    name:str
    age:int 
    email:str

try:
    user = User(
        name="Tejas",
        age=21,
        email="tejas@example.com"
    )

    print(user)

except ValidationError as e:
    print(e)




try:
    user = User(
        name="Tejas",
        age="hello",
        email="tejas@example.com"
    )

    print(user)

except ValidationError as e:
    print(e)
```

# Question 5:
Create a temporary file, write a few lines to it,
read them back, and confirm that the file is cleaned up automatically.

```python
import os
import tempfile

#create temp file
with tempfile.NamedTemporaryFile(mode="w+",delete=True) as temp:
    print("temporary file:",temp.name)

    temp.write("hello python ")
    temp.write("\nthis is temporary file")

    temp.seek(0)

    content = temp.read()

    print("temp content: ", content)

    #check the file exists inside with block
    print("file exists inside the block",os.path.exists(temp.name))
#check file outside the block
print("file exists after the block", os.path.exists(temp.name))
```
