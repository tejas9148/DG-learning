# Q1
Implement the following generator-based solutions: 

(a) fibonacci_gen() that yields Fibonacci numbers indefinitely (no list, no recursion). 

(b) take(n, gen) returning the first n values from any generator as a list. 

(c) squares_gen(limit) yielding squares of integers up to limit.

(d) Use both generators to print the first 10 Fibonacci numbers that are also perfect squares. 

(e) Predict the output: def countdown(n):     while n > 0:         yield n         n -= 1 gen = countdown(5) print(next(gen)) print(next(gen)) print(list(gen)) 

```python


#   (a)

def fibinocci():
    a,b=0,1
    while True:
        yield a
        a,b=b,a+b

gen=fibinocci()
for n in range(10):
    print(next(gen))


# (b)
def take(n,gen):
    result = []
    for i in range(n):
        result.append(next(gen))
    return result
gen = fibinocci()
print(take(5,gen))

# (c)
def squares_gen(limit):
    for i in range(limit+1):
        yield i*i
g=squares_gen(10)
print(list(g))

# (d)

def fib():
    a,b=0,1
    while True:
        yield a
        a,b=b,a+b

def squares(limit):
    for i in range(limit+1):
        yield i**2

fib = fibinocci()
sq=set(squares(1000))
result=[]
for value in fib:
    if value in sq:
        result.append(value)
    if value >1000:
        break
print(result)

# (e)
#predict the output
# 5
# 4
# [3,2,1] 


def countdown(n):
    while n > 0:
        yield n
        n -= 1

gen = countdown(5)
print(next(gen))
print(next(gen))
print(list(gen))

#actual output :
# 5
# 4
# [3,2,1]
```

Topic: JSON & API Interaction 

Write a script interacting with https://jsonplaceholder.typicode.com/posts: 

(a) Fetch all 100 posts. Handle ConnectionError and JSONDecodeError. 

(b) Parse and print title + body of posts 1, 25, 50, 100. 

(c) Filter posts where body contains more than 5 words. Print the count. 

(d) Build {userId: [list of post titles]} and save as posts_by_user.json with indent=2. Add timeout=5. 

(e) Predict the output: import json data = '[{"id":1,"title":"Hello"},{"id":2,"title":"World"}]' posts = json.loads(data) titles = {p['id']: p['title'].upper() for p in posts} print(titles) print(json.dumps(titles)) 





```python
# (a)
import requests
import json

url="https://jsonplaceholder.typicode.com/posts"

try:
    response = requests.get(url , timeout=5)
    response.raise_for_status()
    posts=response.json()
    print("number of posts :", len(posts))
except requests.exceptions.ConnectionError:
    print("connection failed")
except json.JSONDecodeError:
    print("Invalid json response")
print(type(posts))
print(type(posts[0]))

print(posts[0])
```

```python
# (b)
for index in [0,24,49,99]:
    post=posts[index]

    print("post id :",post["id"])
    print("title : " , post["title"])
    print("body : ", post["body"])
    print("\n")
```

```python
# (C)
filtered_post =[]
for post in posts:
    if len(post["body"].split())>5:
        filtered_post.append(post)

print("count: " , len(filtered_post))
```

```python
posts_by_user={}
for post in posts:
    user_id=post["userId"]
    title=post["title"]
    if user_id not in posts_by_user:
        posts_by_user[user_id]=[]
    posts_by_user[user_id].append(title)

print(posts_by_user)

with open("posts_by_user.json","w") as file:
    json.dump(posts_by_user,file,indent=2)
print("file saved sucessfully")
```

```python
# (E)
# predict the output:
#  { 1:'HELLO",2:'WORLD'}
#  { "1":"HELLO","2":"WORLD"}

data = '[{"id":1,"title":"Hello"},{"id":2,"title":"World"}]'

posts = json.loads(data)

titles = {
    p["id"]: p["title"].upper()
    for p in posts
}

print(titles)
print(json.dumps(titles))

#predicted output
# {1: 'HELLO', 2: 'WORLD'}
# {"1": "HELLO", "2": "WORLD"}
```


# Q4

Build a complete ETL pipeline in etl_pipeline.py: 

Extract: Fetch from /users and /posts at jsonplaceholder.typicode.com. 

Transform: (a) Create users DataFrame. (b) Add word_count column to posts. (c) Merge on userId/id. (d) Filter word_count > 30. (e) Add category: Short/Medium/Long. 

Load: Save to output/etl_result.csv. Create directory if missing. 

Add main() and print a summary line with counts. 

(e) Predict the output: import pandas as pd posts = [{'userId':1,'body':'hello world foo bar'},{'userId':1,'body':'hi'},{'userId':2,'body':'a b c d e f g h'}] df = pd.DataFrame(posts) df['wc'] = df['body'].apply(lambda x: len(x.split())) print(df['wc'].tolist()) print(len(df[df['wc'] > 3])) 



```python
import os
import requests
import pandas as pd

url = "https://jsonplaceholder.typicode.com"

def extract():
    try:
        users_response = requests.get(f"{url}/users",timeout=10)
        posts_response = requests.get(f"{url}/posts",timeout=10)

        users_response.raise_for_status()
        posts_response.raise_for_status()

        users = users_response.json()
        posts = posts_response.json()

        return users , posts
    except requests.exceptions.ConnectionError:
        print("error : could not connect to the api")
    except requests.exceptions.Timeout:
        print("error : Api request timed out")
    except requests.exceptions.HTTPError as e:
        print(f"Error : HTTP error occured: {e}")
    except ValueError:
        print("Error:api returned invalid json")

def transform(users , posts):
    # (a)  create users and posts dataframe dataframe
    users_df = pd.DataFrame(users)
    posts_df = pd.DataFrame(posts)

    # (b)  add word_count column
    posts_df["word_count"]=posts_df["body"].apply(lambda x: len(x.split()))

    # (c) merge posts with users on posts.userid = users.id
    merged_df = posts_df.merge(users_df , left_on = "userId", right_on="id",suffixes=("_posts","_user"))

    # (d) filter word_count >30
    filtered_df = merged_df[merged_df["word_count"]>30].copy()
    # (e) add category
    def get_category(word_count):
        if word_count <10:
            return "short"
        elif word_count <=30:
            return "Medium"
        else:
            return "Long"

    filtered_df["category"]=filtered_df["word_count"].apply(get_category)

    return filtered_df

def load(df):

    try:
        os.makedirs("output", exist_ok=True)

        output_path = "output/etl_result.csv"
        df.to_csv(output_path, index=False)

        print(f"Successfully saved to {output_path}")
        return output_path
    except OSError as e:
        print(f"Error creating output directory: {e}")

    except PermissionError:
        print("Error: Permission denied while writing the file.")

    except Exception as e:
        print(f"Error while saving CSV: {e}")

def main():
    users, posts = extract()
    if users is None or posts is None:
        print("etl failed during extraction")
        return 
    result = transform(users, posts)

    output_path = load(result)

    print(
        f"ETL completed: "
        f"{len(users)} users, "
        f"{len(posts)} posts fetched, "
        f"{len(result)} rows loaded → {output_path}"
    )


if __name__ == "__main__":
    main()
```

# Q5

Given a list of at least 6 employee dicts (name, dept, salary) across 3 departments. Use ONLY functional tools: 

(a) List comprehension: names of all 'Engineering' employees. 

(b) Dict comprehension: {name: salary} for employees earning > 60000. 

(c) Set comprehension: unique departments. 

(d) functools.reduce() to calculate total payroll. sorted() with lambda to print top 3 earners. 

(e) Predict the output: from functools import reduce emps = [{'name':'A','dept':'Eng','salary':80000},{'name':'B','dept':'HR','salary':55000},{'name':'C','dept':'Eng','salary':75000}] eng = [e['name'] for e in emps if e['dept']=='Eng'] total = reduce(lambda acc,e: acc+e['salary'], emps, 0) print(eng, total) 



```python
emps = [
    {"name": "Alice", "dept": "Engineering", "salary": 80000},
    {"name": "Bob", "dept": "HR", "salary": 55000},
    {"name": "Charlie", "dept": "Engineering", "salary": 75000},
    {"name": "David", "dept": "Sales", "salary": 65000},
    {"name": "Eva", "dept": "HR", "salary": 70000},
    {"name": "Frank", "dept": "Engineering", "salary": 90000}
]

# (a) list comprehension
engineering_names=[
    e["name"] for e in emps if e["dept"]=="Engineering"
]
print(engineering_names)

# (b) salary > 60000
high_salary = {
    emp["name"] : emp["salary"]
    for emp in emps if emp["salary"]>60000
}
print(high_salary)

# (c) 
departments= {
    emp["dept"] for emp in emps
}
print(departments)

# (d)
from functools import reduce

total_payrolls = reduce(
    lambda acc ,e :acc +e["salary"] , emps , 0
)
print("total payroll :",total_payrolls)

top_3 = sorted(
    emps , key = lambda e : e["salary"],reverse=True
)[:3]
print("top 3 earners :")
for e in  top_3:
    print(e["name"] , e["salary"])


```

```python
# (e)
# predict output:
# ['A' , 'C']  210000
emps = [
    {'name':'A','dept':'Eng','salary':80000},
    {'name':'B','dept':'HR','salary':55000},
    {'name':'C','dept':'Eng','salary':75000}
]

eng = [e['name'] for e in emps if e['dept']=='Eng']

total = reduce(
    lambda acc,e: acc+e['salary'],
    emps,
    0
)

print(eng, total)
# predicted output
# ['A', 'C'] 210000
```
